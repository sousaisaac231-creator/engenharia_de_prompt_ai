# **Calculadora Simples**

```python
try:
    # Início do bloco try para tratamento de erros durante a execução

    # Solicita a entrada do primeiro número real ao usuário
    num1_str = input("Digite o primeiro número real: ")
    # Converte a string de entrada para um número de ponto flutuante (float)
    num1 = float(num1_str)

    # Solicita a entrada do segundo número real ao usuário
    num2_str = input("Digite o segundo número real: ")
    # Converte a string de entrada para um número de ponto flutuante (float)
    num2 = float(num2_str)

    # Apresenta o menu de operações disponíveis ao usuário
    print("\nEscolha a operação:")
    print("1 - Somar")
    print("2 - Subtrair")

    # Solicita a escolha da operação ao usuário
    escolha_str = input("Digite sua escolha (1 ou 2): ")
    # Converte a string de escolha para um número inteiro
    escolha = int(escolha_str)

    # Inicializa variáveis para o resultado e o nome da operação
    resultado = 0
    operacao_nome = ""

    # Verifica a escolha do usuário e realiza a operação correspondente
    if escolha == 1:
        # Executa a operação de soma
        resultado = num1 + num2
        operacao_nome = "Soma"
        # Exibe o nome da operação e o resultado formatado com 2 casas decimais
        print(f"\nOperação: {operacao_nome}")
        print(f"Resultado: {num1} + {num2} = {resultado:.2f}")
    elif escolha == 2:
        # Executa a operação de subtração
        resultado = num1 - num2
        operacao_nome = "Subtração"
        # Exibe o nome da operação e o resultado formatado com 2 casas decimais
        print(f"\nOperação: {operacao_nome}")
        print(f"Resultado: {num1} - {num2} = {resultado:.2f}")
    else:
        # Caso a escolha seja inválida, informa o usuário
        print("\nEscolha inválida. Por favor, digite 1 para Somar ou 2 para Subtrair.")

# Bloco except para tratar erros de ValueError (ex: entrada não numérica para float ou int)
except ValueError:
    print("\nEntrada inválida. Certifique-se de digitar números reais para os operandos e um número inteiro para a escolha da operação.")
# Bloco except genérico para capturar qualquer outro tipo de erro inesperado
except Exception as e:
    print(f"Ocorreu um erro inesperado: {e}")
```
